```python
import numpy as np
import matplotlib.pyplot as plt


clicks_perSecond=np.arange(18)
ObservedCounts=[1,6,11,28,56,105,126,146,164,161,123,101,74,53,23,15,9,5]
```

```python
plt.figure(figsize=(10,10))
plt.plot(clicks_perSecond,ObservedCounts)
plt.title("Distribution of the data")
plt.ylabel("Number of clicks per second")
plt.xlabel("Number of bins")
```

```python
from scipy.stats import poisson
sample_mean=np.average(clicks_perSecond)
poissonDistribution=poisson.pmf(clicks_perSecond,mu=sample_mean)*ObservedCounts

plt.plot(clicks_perSecond,poissonDistribution,marker='o',linestyle='--',color='red')
plt.plot(clicks_perSecond,ObservedCounts,marker='*',linestyle='-',color='green')
```
